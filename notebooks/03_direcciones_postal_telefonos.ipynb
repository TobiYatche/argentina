## Direcciones — parser local (sin red)

````python
d = "Av. Santa Fe 3253 Piso 2 Depto B"
arg.direcciones.normalizar(d)
````

````python
arg.direcciones.parsear(d)
````

````python
arg.direcciones.extraer_calle(d), arg.direcciones.extraer_altura(d)
````

> Para georreferenciación real usar `arg.geo.direcciones` (extra `[georef]`).

## Códigos postales

Soporta los dos formatos vigentes: CP4 (4 dígitos) y CPA (8 chars).

````python
arg.postal.validar_cp4("1425")
````

````python
arg.postal.validar_cpa("C1425ABC")
````

````python
arg.postal.extraer_cp4("C1425ABC")
````

````python
arg.postal.provincia_por_cpa("X5000AAA")
````

````python
arg.postal.validar_cpa_provincia("X5000AAA", "Córdoba")
````

## Teléfonos

````python
arg.telefonos.limpiar("+54 9 11 1234-5678")
````

````python
arg.telefonos.validar("+54 9 11 1234-5678")
````

````python
arg.telefonos.normalizar_e164("011 4321-1234")
````

````python
arg.telefonos.es_celular("+54 9 351 1234567")
````

````python
arg.telefonos.extraer_caracteristica("+54 9 351 1234567")
````

### Extraer teléfonos de texto libre

````python
arg.telefonos.extraer_de_texto("Llamame al 11 1234-5678 o al 0351 7654321")
````

````python
arg.telefonos.extraer_de_texto("Tel: +54 9 11 1234-5678", normalizar=True)
````

## Identificador universal

Cuando no sabés qué tenés en una columna, `arg.identificar` deduce el tipo.

````python
for v in ["C1425ABC", "1425", "+54 9 11 1234-5678", "Av Santa Fe 3253"]:
    print(v, "→", arg.identificar(v))
````

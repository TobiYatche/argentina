# argentina.formato — Pruebas interactivas

Punto único de entrada para formatear datos argentinos para mostrar. Agrupa funciones nuevas (teléfono, pesos, código postal, fecha) y reexporta las `formatear_*` que ya viven en otros módulos.

## 1. Setup

```python
import argentina as arg
from datetime import date

print(f"argentina v{arg.__version__}")
```

## 2. Teléfono

Tres estilos: `nacional` (default), `e164`, `internacional`.

```python
arg.formato.telefono("1140404040")
```

```python
arg.formato.telefono("1140404040", estilo="e164")
```

```python
# Celular: agrega el 9 de movilidad
arg.formato.telefono("+5491140404040", estilo="internacional")
```

```python
# Característica de 3 dígitos (Córdoba 351)
arg.formato.telefono("3514445555")
```

```python
# Inválido → None
print(arg.formato.telefono("xyz"))
```

## 3. Pesos

Convención argentina: `.` para miles, `,` para decimales.

```python
arg.formato.pesos(1_500_000)
```

```python
arg.formato.pesos(1_500_000.5, decimales=2)
```

```python
# Negativos: el signo va antes del símbolo
arg.formato.pesos(-1000)
```

```python
# Sin símbolo
arg.formato.pesos(100, simbolo="")
```

## 4. Código postal

Distingue CP4 (4 dígitos) de CPA (1 letra + 4 dígitos + 3 letras).

```python
arg.formato.codigo_postal("1414")
```

```python
# Normaliza mayúsculas y limpia separadores
arg.formato.codigo_postal("  c1414-baa ")
```

## 5. Fecha

Estilos: `corto` (default, dd/mm/aaaa), `largo` (en español), `iso`.

```python
arg.formato.fecha(date(2026, 5, 13))
```

```python
arg.formato.fecha(date(2026, 5, 13), estilo="largo")
```

```python
# Acepta strings con formatos comunes
arg.formato.fecha("2026-05-13", estilo="largo")
```

## 6. Reexports

Las funciones canónicas siguen en `personas`, `bancos` y `patentes`. `formato` agrega nombres descubribles que delegan al original.

```python
arg.formato.dni("12345678")
```

```python
arg.formato.cuit("20123456786")
```

```python
arg.formato.cbu("2850590940090418135201")
```

```python
arg.formato.patente("AB123CD")
```

## 7. Patrón de uso: panel resumido

Mostrar varios campos de un registro con el formato canónico.

```python
registro = {
    "dni": "12345678",
    "cuit": "20123456786",
    "telefono": "1140404040",
    "cp": "C1414BAA",
    "fecha": "2026-05-13",
    "monto": 1_234_567.89,
}

print("DNI      :", arg.formato.dni(registro["dni"]))
print("CUIT     :", arg.formato.cuit(registro["cuit"]))
print("Teléfono :", arg.formato.telefono(registro["telefono"]))
print("CP       :", arg.formato.codigo_postal(registro["cp"]))
print("Fecha    :", arg.formato.fecha(registro["fecha"], estilo="largo"))
print("Monto    :", arg.formato.pesos(registro["monto"], decimales=2))
```

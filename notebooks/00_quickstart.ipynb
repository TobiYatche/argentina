## 2. Provincias y departamentos

Lookup flexible: nombre, código INDEC, ISO 3166-2 o alias.

````python
arg.provincias.lookup("PBA")
````

````python
arg.provincias.lookup("Córdoba")  # con tilde
````

````python
arg.provincias.lookup("14")  # código INDEC
````

````python
p = arg.provincias.CORDOBA
p.nombre, p.capital, p.region, p.poblacion_2022
````

````python
arg.departamentos.lookup("06427")  # La Matanza
````

````python
len(arg.departamentos.por_provincia("PBA"))  # 135 partidos
````

## 7. Formato canónico

````python
arg.formato.cuit("20123456786")
````

````python
arg.formato.pesos(1_500_000)
````

## 9. Ajustar montos por IPC (offline)

Snapshot embebido en stdlib pura. Para series largas online ver `arg.economia` (extra `[economia]`).

````python
from datetime import date
arg.indices.ajustar_ipc(10_000, desde=date(2017,1,1), hasta=date(2024,1,1))
````

````python
arg.indices.disponibles()
````

## Parseo flexible

`parsear_fecha` acepta los formatos más comunes en bases argentinas.

````python
arg.fechas.parsear_fecha("31/12/2024")
````

````python
arg.fechas.parsear_fecha("2024-12-31")
````

````python
arg.fechas.fecha_iso("31/12/2024")
````

````python
arg.fechas.es_fecha_valida("31/12/2024")
````

## Formato canónico

`arg.formato.fecha` cubre los tres estilos típicos.

````python
from datetime import date
arg.formato.fecha(date(2026, 5, 13))               # corto
````

````python
arg.formato.fecha("13/05/2026", estilo="largo")
````

````python
arg.formato.fecha("13/05/2026", estilo="iso")
````

## Índices económicos offline

`arg.indices` provee IPC nacional, UVA, CER e ICL como snapshot embebido (stdlib pura, sin red).

````python
arg.indices.disponibles()
````

````python
from datetime import date
arg.indices.cobertura("ipc_nacional")
````

````python
arg.indices.ipc(date(2024, 1, 1)), arg.indices.uva(date(2024, 1, 1))
````

### Ajustar montos por inflación

````python
arg.indices.ajustar_ipc(
    10_000,
    desde=date(2017, 1, 1),
    hasta=date(2024, 1, 1),
)
````

````python
# Solo el factor (reutilizable)
arg.indices.factor_ipc(desde=date(2017, 1, 1), hasta=date(2024, 1, 1))
````

````python
# Genérico con cualquier índice
arg.indices.ajustar("uva", 1000, desde=date(2020, 1, 1), hasta=date(2024, 1, 1))
````

Si pedís una fecha fuera del rango cubierto, levanta `ValueError` con la cobertura disponible. Nunca extrapola silenciosamente.

## Monedas históricas por fecha

````python
arg.monedas.en(date(1986, 6, 1))  # Austral
````

````python
arg.monedas.actual()  # ARS
````

````python
arg.monedas.convertir(10_000, "₳")  # nominal, no inflación
````

> Para deflactar/inflar usando IPC oficial, usar `arg.indices.ajustar_ipc(...)`. `arg.monedas.convertir` solo sigue cambios nominales de moneda.

## DNI

`limpiar_dni` saca todo lo que no sea dígito. `validar_dni` chequea 7-8 dígitos. `formatear_dni` agrega puntos.

````python
arg.personas.limpiar_dni("12.345.678")
````

````python
arg.personas.validar_dni("12345678")
````

````python
arg.personas.formatear_dni("12345678")
````

## CUIT / CUIL

Valida **dígito verificador real** (algoritmo oficial, no solo largo).

````python
arg.personas.validar_cuit("20-12345678-6")  # dígito correcto
````

````python
arg.personas.validar_cuit("20-12345678-3")  # dígito incorrecto
````

````python
arg.personas.calcular_digito_cuit("2012345678")  # → '6'
````

````python
arg.personas.tipo_cuit("20-12345678-6")  # persona_fisica
````

````python
arg.personas.tipo_cuit("30-71234567-1")  # persona_juridica
````

````python
arg.personas.extraer_dni_de_cuit("20-12345678-6")
````

````python
arg.personas.formatear_cuit("20123456786")
````

## Nombres

Normalización canónica + helpers para primer nombre / apellido.

````python
arg.personas.normalizar_nombre(" María   Laura ")
````

````python
arg.personas.primer_nombre("María Laura")
````

````python
arg.personas.apellido_principal("Pérez Gómez")
````

## Estimar año de nacimiento desde el DNI

Modelo lineal calibrado contra hitos del RENAPER. Margen típico ±2-3 años.

````python
arg.personas.estimar_año_nacimiento("35.000.000")
````

````python
arg.personas.estimar_dni(1990)
````

Casos borde: la franja 60.000.000–69.999.999 está reservada para CUIT de extranjeros, no son DNIs reales.

````python
arg.personas.estimar_año_nacimiento(65_000_000)  # → None
````

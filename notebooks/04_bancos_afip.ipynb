## Monotributo

`arg.afip.monotributo_*` expone las categorías oficiales por año.

````python
arg.afip.monotributo_anios_disponibles()
````

````python
# Listar todas las categorías del año
cats = arg.afip.monotributo_categorias(anio=2024)
for c in cats:
    print(c.letra, f"${c.facturacion_max:,}".replace(",", "."))
````

````python
# Categoría según facturación anual
arg.afip.monotributo_categoria_por_facturacion(12_000_000, anio=2024)
````

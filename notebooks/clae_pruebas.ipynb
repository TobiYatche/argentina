# argentina.clae — Pruebas interactivas

Clasificador de Actividades Económicas AFIP. Subset representativo (~120 códigos).

```python
import argentina as arg
print(f"argentina v{arg.__version__}, {len(arg.clae)} actividades embebidas")
```

## Lookup por código

```python
arg.clae.lookup("620100")
```

```python
arg.clae.es_valido("620100"), arg.clae.es_valido("999999")
```

## Filtros por sector y grupo

```python
for a in arg.clae.por_sector("J"):
    print(a.codigo, a.descripcion[:60])
```

```python
[a.codigo for a in arg.clae.por_grupo("6201")]
```

## Buscar por descripción

```python
for a in arg.clae.buscar("informática"):
    print(a.codigo, a.descripcion[:60])
```

## Sectores disponibles

```python
for s in arg.clae.sectores():
    print(s.letra, s.nombre)
```

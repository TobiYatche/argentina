## Distancia haversine (sin extras)

`arg.geo.distancia` usa solo `math` de stdlib.

````python
arg.geo.distancia("Buenos Aires", "Córdoba")  # ~647 km
````

````python
arg.geo.distancia("Buenos Aires", "Ushuaia")  # ~2373 km
````

````python
arg.geo.distancia((-34.6, -58.4), (-31.4, -64.2))
````

## Coordenadas universales

`arg.coordenadas` resuelve `(lat, lon)` desde cualquier identificador.

````python
arg.coordenadas("Rosario")
````

````python
arg.coordenadas("EZE")  # aeropuerto
````

````python
arg.coordenadas(arg.provincias.MENDOZA)
````

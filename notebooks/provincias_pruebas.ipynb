## 1. Setup e imports

## 3. Campos de cada provincia

`Provincia` es un dataclass frozen con 5 campos.

````python
p = arg.provincias.CORDOBA
print("nombre:      ", p.nombre)
print("codigo_indec:", p.codigo_indec)
print("iso_id:      ", p.iso_id)
print("region:      ", p.region)
print("capital:     ", p.capital)
````

````python
# frozen: no se puede mutar
try:
    arg.provincias.CORDOBA.nombre = "otro"
except Exception as e:
    print(type(e).__name__, '→', e)
````

## 4. Lookup flexible

`lookup` acepta nombre (con o sin tildes), código INDEC, ID ISO 3166-2, o aliases comunes. Es case-insensitive.

````python
# Por nombre con tilde
arg.provincias.lookup("Córdoba")
````

````python
# Por nombre sin tilde, minúsculas
arg.provincias.lookup("cordoba")
````

````python
# Por código INDEC
arg.provincias.lookup("14")
````

````python
# Por código INDEC sin padding ("6" en lugar de "06")
arg.provincias.lookup("6"), arg.provincias.lookup("06")
````

````python
# Por ID ISO 3166-2
arg.provincias.lookup("AR-X")
````

````python
# El ISO también funciona en minúsculas
arg.provincias.lookup("ar-x")
````

## 6. Casos borde

Valores que no matchean devuelven `None`.

````python
print(arg.provincias.lookup("Patagonia"))   # región, no provincia
print(arg.provincias.lookup("99"))          # código INDEC inexistente
print(arg.provincias.lookup("AR-ZZ"))       # ISO inválido
print(arg.provincias.lookup(""))            # string vacío
print(arg.provincias.lookup(None))          # None
````

## 7. Iterar sobre todas las provincias

`listar()` devuelve una tupla inmutable de las 24.

````python
for p in arg.provincias.listar():
    print(f"{p.codigo_indec}  {p.iso_id}  {p.region:10s}  {p.nombre}")
````

````python
# Agrupar por región
from collections import defaultdict

por_region = defaultdict(list)
for p in arg.provincias.listar():
    por_region[p.region].append(p.nombre)

for region, nombres in sorted(por_region.items()):
    print(f"{region}: {', '.join(nombres)}")
````

````python
# Capitales
for p in arg.provincias.listar():
    print(f"{p.nombre:35s} → {p.capital}")
````
